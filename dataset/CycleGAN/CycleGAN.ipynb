# Process data

### SEN12MSCR
- Change the name of the `folder_name` accordingly. (`SEN12MSCR_test_translated_full`, `SEN12MSCR_test_translated_partial`, etc.)
- Fill in the path for the `data_root`.

```python
import shutil
import os
from tqdm import tqdm
from pathlib import Path

# Define source and destination directories
folder_name = 'SEN12MSCR_test_translated_partial'

data_root = '../SEN12MS-CR_1'

source_dir = './results/cyclegansen12/' + folder_name
ori_images_dir = Path(source_dir) / 'images'
destination_dir_1 = Path(source_dir) / (os.path.basename(source_dir) + '_npy')

# Create destination directory if it doesn't exist
os.makedirs(destination_dir_1, exist_ok=True)

# Rename files ending with _fake_A.npy to .npy with progress tracking
for file_name in tqdm(os.listdir(ori_images_dir), desc="Renaming files"):
    if file_name.endswith('_fake_A.npy'):
        new_name = file_name.replace('_fake_A.npy', '.npy')
        os.rename(os.path.join(ori_images_dir, file_name), os.path.join(ori_images_dir, new_name))

# Rename folder to destination_dir

ori_images_dir.rename(destination_dir_1)
print("Renaming completed. All files have been renamed and moved to the new directory.")

destination_dir_2 = data_root + '/' + os.path.basename(destination_dir_1)
print(f"Moving folder from {os.path.basename(destination_dir_1)} to {os.path.basename(destination_dir_2)}...")
shutil.move(destination_dir_1, destination_dir_2)
print(f"Folder moved to {destination_dir_2}")
```
